```python
# FINAL ASSIGNMENT - VISUALIZATIONS USING MATPLOTLIB, SEABORN & FOLIUM

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import folium

%matplotlib inline
sns.set_style("whitegrid")

# --------------------------------------------------
# 1. CREATE DATASET
# --------------------------------------------------

data = {
    "City": ["Delhi", "Mumbai", "Bangalore", "Chennai", "Kolkata", "Hyderabad"],
    "Population": [19000000, 21000000, 12000000, 11000000, 15000000, 10000000],
    "Average_Temperature": [25, 28, 24, 29, 27, 26],
    "Rainfall": [800, 2200, 970, 1400, 1800, 900],
    "Latitude": [28.6139, 19.0760, 12.9716, 13.0827, 22.5726, 17.3850],
    "Longitude": [77.2090, 72.8777, 77.5946, 80.2707, 88.3639, 78.4867]
}

df = pd.DataFrame(data)

print("DATASET")
display(df)

# --------------------------------------------------
# 2. MATPLOTLIB - BAR CHART
# --------------------------------------------------

plt.figure(figsize=(10, 5))
plt.bar(df["City"], df["Population"])
plt.title("Population of Major Indian Cities")
plt.xlabel("City")
plt.ylabel("Population")
plt.xticks(rotation=45)
plt.show()

# --------------------------------------------------
# 3. MATPLOTLIB - LINE CHART
# --------------------------------------------------

plt.figure(figsize=(10, 5))
plt.plot(
    df["City"],
    df["Average_Temperature"],
    marker="o"
)
plt.title("Average Temperature by City")
plt.xlabel("City")
plt.ylabel("Temperature (°C)")
plt.xticks(rotation=45)
plt.show()

# --------------------------------------------------
# 4. MATPLOTLIB - SCATTER PLOT
# --------------------------------------------------

plt.figure(figsize=(8, 5))
plt.scatter(
    df["Average_Temperature"],
    df["Rainfall"]
)
plt.title("Temperature vs Rainfall")
plt.xlabel("Average Temperature (°C)")
plt.ylabel("Rainfall (mm)")
plt.show()

# --------------------------------------------------
# 5. MATPLOTLIB - PIE CHART
# --------------------------------------------------

plt.figure(figsize=(8, 8))
plt.pie(
    df["Population"],
    labels=df["City"],
    autopct="%1.1f%%"
)
plt.title("Population Distribution")
plt.show()

# --------------------------------------------------
# 6. MATPLOTLIB - HISTOGRAM
# --------------------------------------------------

plt.figure(figsize=(8, 5))
plt.hist(
    df["Average_Temperature"],
    bins=5
)
plt.title("Distribution of Average Temperature")
plt.xlabel("Temperature (°C)")
plt.ylabel("Frequency")
plt.show()

# --------------------------------------------------
# 7. SEABORN - BAR PLOT
# --------------------------------------------------

plt.figure(figsize=(10, 5))
sns.barplot(
    x="City",
    y="Rainfall",
    data=df
)
plt.title("Rainfall by City")
plt.xlabel("City")
plt.ylabel("Rainfall (mm)")
plt.xticks(rotation=45)
plt.show()

# --------------------------------------------------
# 8. SEABORN - SCATTER PLOT
# --------------------------------------------------

plt.figure(figsize=(8, 5))
sns.scatterplot(
    x="Average_Temperature",
    y="Rainfall",
    data=df,
    s=100
)
plt.title("Temperature vs Rainfall")
plt.xlabel("Average Temperature (°C)")
plt.ylabel("Rainfall (mm)")
plt.show()

# --------------------------------------------------
# 9. SEABORN - BOX PLOT
# --------------------------------------------------

plt.figure(figsize=(8, 5))
sns.boxplot(
    y=df["Average_Temperature"]
)
plt.title("Temperature Distribution")
plt.ylabel("Temperature (°C)")
plt.show()

# --------------------------------------------------
# 10. SEABORN - PAIR PLOT
# --------------------------------------------------

sns.pairplot(
    df[
        [
            "Population",
            "Average_Temperature",
            "Rainfall"
        ]
    ]
)
plt.show()

# --------------------------------------------------
# 11. SEABORN - CORRELATION HEATMAP
# --------------------------------------------------

plt.figure(figsize=(8, 6))

correlation = df[
    [
        "Population",
        "Average_Temperature",
        "Rainfall"
    ]
].corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap")
plt.show()

# --------------------------------------------------
# 12. FOLIUM - CREATE MAP
# --------------------------------------------------

india_map = folium.Map(
    location=[22.5, 78.9],
    zoom_start=5
)

# --------------------------------------------------
# 13. ADD CITY MARKERS
# --------------------------------------------------

for _, row in df.iterrows():

    folium.Marker(
        location=[
            row["Latitude"],
            row["Longitude"]
        ],
        popup=(
            f"City: {row['City']}<br>"
            f"Population: {row['Population']:,}<br>"
            f"Temperature: {row['Average_Temperature']} °C<br>"
            f"Rainfall: {row['Rainfall']} mm"
        ),
        tooltip=row["City"]
    ).add_to(india_map)

# Display map
display(india_map)

# --------------------------------------------------
# 14. SAVE MAP
# --------------------------------------------------

india_map.save("india_cities_map.html")

print("All visualizations completed successfully!")
print("Folium map saved as: india_cities_map.html")
```
